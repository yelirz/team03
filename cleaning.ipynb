```python
#make a copy of the original dataset
cleanprice=price.copy()
#convert all prices in strings into numbers
#select columns that are not the first four rows,put NaN in :
for col in cleanprice.columns[4:]:
  cleanprice[col]=pd.to_numeric(cleanprice[col],errors='coerce')
cleanprice[col].describe()
```

```python
#check if conversion is successful
cleanprice.info()
```

```python
#what kind of data is each column
cleanprice.info(verbose=True, show_counts=True)
```

```python
#check the number of blanks
cleanprice.isna().sum()
```

```python
#replace all the blanks with median price of each respective year
for col in cleanprice.columns[4:]:
  cleanprice[col]=cleanprice[col].fillna(cleanprice[col].median())
```

```python
#check if replacement works, that is, if there are still blank spaces
cleanprice.isna().sum()
```

```python
#save the clean copy
cleanprice.to_csv("cleanprice.csv",index=False)
print("Saved",len(price),"rows to cleanprice.csv")
```
